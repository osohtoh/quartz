```python
with open("./tests/sample_data/FX01.pkl", "rb") as f:
    mols = pickle.load(f)
```

```python
# mols[0].res.keys()
```

```python
with open("tests/sample_data/FX_data.csv", "r", encoding="utf-8-sig") as f:
    csvreader = csv.DictReader(f)
    for row in csvreader:
        # print(row)
        for mol in mols:
            if mol.filename[0:4] is not None and mol.filename[0:4] == row["PDB ID "]:
                mol.pdb_code = row["PDB ID "]
                mol.ki = float(row["Ki (nM)"])
                mol.pki = float(row["pKi "])
                # print(f"Assigned PDB code {row['PDB ID ']} to {mol.filename}")
                break

mols.sort(key=lambda x: x.ki)
```

```python
for i in mols:
    assign_props(
        i,
        [
            "coordination number",
            "dispersion coef",
            "polarizibility",
            "charges",
            "fullpop",
            "fod_s",
            "fod_p",
            "fod_d",
            "fukui+",
            "fukui-",
            "fukui0",
        ],
    )
```

```python
# np.min(np.abs(mols[3].res['coordination number']))
```

```python
grids = qg.create_grid(
    mols,
    props=[
        "coordination number",
        "dispersion coef",
        "polarizibility",
        "charges",
        "fullpop",
        "fod_s",
        "fod_p",
        "fod_d",
        "fukui+",
        "fukui-",
        "fukui0",
    ],
    padding=5.0,
    size=0.3,
    threshold={
        "dispersion coef": 0.1,
        "coordination number": 0.01,
        "fukui+": 0.001,
        "fukui-": 0.001,
        "fukui0": 0.001,
        "polarizibility": 0.01,
        "charges": 0.001,
        "fullpop": 0.01,
        "fod_s": 0.01,
        "fod_p": 0.01,
        "fod_d": 0.01,
    },
    clamp_dist=0.5,
    clamp_falloff=1000,
)
```

```python
overlaps = []
for i in range(2, len(mols)):
    overlaps.append(
        calculate_overlap(
            grids[0:i],
            [
                "coordination number",
                "dispersion coef",
                "polarizibility",
                "charges",
                "fullpop",
                "fod_s",
                "fod_p",
                "fod_d",
                "fukui+",
                "fukui-",
                "fukui0",
            ],
        )
    )
```
